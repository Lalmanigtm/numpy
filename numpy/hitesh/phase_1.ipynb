## Numpy array and basics

```python
import numpy as np 
```

### creating array from list

```python
arr_1d = np.array([1,2,3,4,5])
print(arr_1d)

arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D_array:",arr_2d )
```

## python list VS numpy array

```python
# Example 1: 
py_list = [1,2,3,4,5]
print("python list :", py_list)

np_array = np.array([1,2,3,4,5]) # element wise multiplication
print("numpy array :", np_array * 2)


# Example 2: 
import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n list operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n Numpy operation time : ", time.time() - start)
```

## Creating arrays from scratch

```python
zeros_array = np.zeros((2,3))
print("Zeros array : \n", zeros_array)

ones_array = np.ones((2,3))
print("Ones Array :\n", ones_array)


full_array = np.full((2,3),9)
print("Full Array :\n", full_array)

random_array = np.random.random((2,3))
print("Random Array :\n", random_array)

sequence_array = np.arange(0,20,3)
print("sequence array : \n", sequence_array)
```

## Vector, Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector:\n", vector)

matrices = np.array([[1,2],
                [3,4]])
print("Matrices:\n", matrices) 

tensor = np.array([[[1,2],[3,4]], [[5,6],[7,8]]])
print("Tensor :\n", tensor)
```

## Properties of arrays

```python
arr = np.array([[1,2,3],
                [4,5,6]])

print("Shape : \n", arr.shape)  
print("Size : \n", arr.size)
print("Dimension : \n", arr.ndim)
print("Data type : \n", arr.dtype)
print("Shape : \n", arr.shape)              
```

## Array reshaping

```python
arr = np.arange(12)
print("Original array :", arr)

reshaped = arr.reshape((3,4))
print("reshaped array :", reshaped)
```
